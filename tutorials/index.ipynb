# Tutorials

Run these notebooks to explore simgeometry with real grids and shapes. Each page includes its source notebook for download.

1. [Grids and coordinates](/simgeometry/tutorials/grids-and-coordinates/) explains node locations, masks, and periodic axes.
2. [Shapes and node masks](/simgeometry/tutorials/shapes-and-node-masks/) selects grid nodes, tests point batches, and plots a circle.
3. [Parallel and multipatch grids](/simgeometry/tutorials/parallel-and-multipatch-grids/) covers decomposition metadata and coordinate conversion across patches.
