# Tutorials

Run these notebooks to explore simgeometry with real grids and shapes. Each page includes its source notebook for download.

1. [Grids and coordinates](/simgeometry/tutorials/grids-and-coordinates/) explains node locations, masks, and periodic axes.
2. [Shapes and node masks](/simgeometry/tutorials/shapes-and-node-masks/) selects grid nodes, tests point batches, and plots a circle.
3. [Parallel grids](/simgeometry/tutorials/parallel-grids/) covers MPI decomposition metadata, local index ranges, and neighbouring ranks.

4. [Inspecting geometry](/simgeometry/tutorials/inspecting-geometry/) covers plotting, 3D surfaces and interactive clipping.
5. [Slices and coordinate sampling](/simgeometry/tutorials/slices-and-coordinate-sampling/) covers labelled evaluation and arbitrary planes.
6. [Building composite shapes](/simgeometry/tutorials/building-composite-shapes/) combines, transforms and names objects.
7. [Grid intersections and cut edges](/simgeometry/tutorials/grid-intersections-and-cut-edges/) covers surface crossings and cell fractions.
8. [Sampling, export and diagnostics](/simgeometry/tutorials/sampling-export-and-diagnostics/) covers point clouds, portable meshes and resolution checks.
9. [Boundary outlines](/simgeometry/tutorials/boundary-outlines/) turns 2D shapes into vertex loops and panels for boundary-element solvers, and finds where particle steps cross a boundary.

See [Building the docs](/simgeometry/development/docs/) to add a tutorial.
